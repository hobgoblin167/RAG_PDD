```python
def parse_chapters(rule_text):
    chapter_pattern = re.compile(r'(?m)^(\d{1,2})\.\s+([А-ЯЁ][^\n]*)$')
    return [(m.start(), f"{m.group(1)}. {m.group(2).strip()}")
            for m in chapter_pattern.finditer(rule_text)]

def find_chapter(offset, chapters):
    current = None
    for pos, title in chapters:
        if pos <= offset:
            current = title
        else:
            break
    return current
rule_text=text
chapters = parse_chapters(rule_text)   # rule_text — исходный текст секции с правилами

for c in chunks:
    pos = rule_text.find(c["text"][:30])   # ищем позицию чанка в исходном тексте
    c["chapter"] = find_chapter(pos, chapters)
```

```python
import re

# ------------------------------------------------------------
# 1. Находим позиции разделов в общем тексте
# ------------------------------------------------------------
pos_app1 = text.find("Приложение 1")
pos_app2 = text.find("Приложение 2")
# Основные положения находятся после приложений (после "Приложение N 3")
pos_base = text.find("Приложение N 3")


print(f"Позиции: Приложение1={pos_app1}, Приложение2={pos_app2}, Основные положения={pos_base}")
```

```python

# ------------------------------------------------------------
# 2. Отфильтровываем из исходного chunks только основные правила (до Приложения 1)
#    и термины (они уже добавлены отдельно и находятся до pos_app1)
# ------------------------------------------------------------
main_chunks = []
for c in chunks:
    # Ищем позицию первого вхождения текста чанка в общем тексте
    # (берём первые 50 символов для надёжности)
    sample = c['text'][:50]
    pos = text.find(sample)
    if pos != -1 and pos < pos_app1:
        main_chunks.append(c)
    # Если это термины (point is None и есть 'term'), они тоже должны быть в основных
    elif c.get('term') is not None:
        main_chunks.append(c)

print(f"Основных чанков (правила + термины): {len(main_chunks)}")

# ------------------------------------------------------------
# 3. Парсинг Приложения 1 (дорожные знаки)
# ------------------------------------------------------------
def parse_appendix(text, start_marker, end_marker=None, num_pattern=r'^(?P<num>[\d.]+(?:-[\.\d]+)?)\s+[—–-]?\s*'):
    """Универсальный парсер для приложений по номерам в начале строк."""
    start = text.find(start_marker)
    if start == -1:
        return []
    if end_marker:
        end = text.find(end_marker, start)
        if end == -1:
            end = len(text)
    else:
        end = len(text)
    sub = text[start:end]
    pattern = re.compile(num_pattern, re.MULTILINE)
    matches = list(pattern.finditer(sub))
    chunks = []
    for i, m in enumerate(matches):
        start_pos = m.start()
        end_pos = matches[i+1].start() if i+1 < len(matches) else len(sub)
        chunk_text = sub[start_pos:end_pos].strip()
        num = m.group('num')
        if start_marker.strip()=="Приложение 2":
          chunks.append({
              'point': num,
              'text': chunk_text,
              'section': "Дорожная разметка и ее характеристики"
          })
        else:
          chunks.append({
              'point': num,
              'text': chunk_text,
              'section': "Дорожные знаки"
          })

    return chunks

# Парсим знаки (Приложение 1) до начала Приложения 2
sign_chunks = parse_appendix(text, "Приложение 1", "Приложение 2")
print(f"Знаков: {len(sign_chunks)}")

# ------------------------------------------------------------
# 4. Парсинг Приложения 2 (дорожная разметка)
# ------------------------------------------------------------
# Разметка начинается после "1. Горизонтальная разметка" и т.д.
# Но мы можем взять весь раздел до "Основные положения"
marking_chunks = parse_appendix(text, "Приложение 2", "Приложение N 3")
print(f"Разметка: {len(marking_chunks)}")

# ------------------------------------------------------------
# 5. Парсинг Основных положений (включая перечень неисправностей)
# ------------------------------------------------------------
# Для основных положений номера пунктов - целые числа (1., 2., ... 21.)
# Подпункты оставляем внутри текста
base_start = text.find("Приложение N 3")
base_text = text[base_start:]

# Ищем номера вида "1.", "2.", ... в начале строки (без точек внутри)
pattern_base = re.compile(r'^(?P<num>\d{1,2})\.\s+', re.MULTILINE)
matches = list(pattern_base.finditer(base_text))
base_chunks = []
for i, m in enumerate(matches):
    start_pos = m.start()
    end_pos = matches[i+1].start() if i+1 < len(matches) else len(base_text)
    chunk_text = base_text[start_pos:end_pos].strip()
    num = m.group('num')
    base_chunks.append({
        'point': num,
        'text': chunk_text,
        'section': 'Основные положения по допуску транспортных средств к эксплуатации'
    })

print(f"Основных положений: {len(base_chunks)}")

# ------------------------------------------------------------
# 6. Объединяем все списки
# ------------------------------------------------------------
final_chunks = main_chunks + sign_chunks + marking_chunks + base_chunks
print(f"Итоговое количество чанков: {len(final_chunks)}")

# (Опционально) Сохраняем для дальнейшего использования
# import pickle
# with open("/content/final_chunks.pkl", "wb") as f:
#     pickle.dump(final_chunks, f)
```

# Настройка пайплайна

```python
def retrieve(query, index, chunks, embedding_model, top_k=3):
    """Находит top_k наиболее релевантных чанков для запроса."""
    query_embedding = embedding_model.encode([query], normalize_embeddings=True)
    distances, indices = index.search(query_embedding.astype('float32'), top_k)

    retrieved = []
    for i, idx in enumerate(indices[0]):
        if idx != -1:
            chunk = chunks[idx]
            retrieved.append({
                'chunk': chunk,
                'score': distances[0][i],
                'text': chunk['text']
            })
    return retrieved

# Тест поиска
test_query = "какие документы должен иметь водитель"
results = retrieve(test_query, index, final_chunks, embedding_model, top_k=3)

print(f"Запрос: {test_query}\n")
for i, res in enumerate(results):
    chunk = res['chunk']
    print(f"Результат {i+1} (оценка: {res['score']:.4f}):")
    if chunk.get('term'):
        print(f"  Термин: {chunk['term']}")
    elif chunk.get('chapter'):
        print(f"  Глава: {chunk['chapter']}, пункт: {chunk['point']}")
    elif chunk.get('section'):
        print(f"  Раздел: {chunk['section']}, пункт: {chunk['point']}")
    print(f"  Текст: {res['text'][:200]}...")
    print("-" * 40)
```

```python
def generate_answer(query, retrieved_chunks, llm):
    """Генерирует ответ на основе найденных чанков."""
    # Формируем контекст
    context_parts = []
    for i, res in enumerate(retrieved_chunks):
        chunk = res['chunk']

        # Формируем заголовок в зависимости от типа чанка
        if chunk.get('term'):
            header = f"Термин \"{chunk['term']}\""
        elif chunk.get('chapter'):
            header = f"{chunk['chapter']} (пункт {chunk['point']})"
        elif chunk.get('section'):
            header = f"{chunk['section']} (пункт {chunk['point']})"
        else:
            header = "Информация"

        context_parts.append(f"{header}:\n{chunk['text']}")

    context = "\n\n".join(context_parts)

    system_prompt = """
Ты — эксперт по правилам дорожного движения (ПДД) Российской Федерации.
Правила ответа:
1. Используй ТОЛЬКО информацию из предоставленного контекста.
2. Если в контексте нет ответа — скажи "В предоставленном контексте нет информации об этом".
3. Отвечай подробно и развернуто, объясняя все нюансы.
4. Всегда указывай конкретные пункты ПДД (например, "пункт 1.3 ПДД").
5. Если есть исключения — обязательно их перечисли

Важно! Различай:
- Пункты ПДД (нумерация 1.1, 1.2... в основной части)
- Номера дорожной разметки (1.1, 1.2... в Приложении 2)

При ответе о разметке:
- Сплошные линии (1.1, 1.2, 1.3): пересекать запрещено
- Прерывистые линии (1.5, 1.6, 1.7, 1.8): пересекать разрешено
- Комбинированная (1.11): со стороны прерывистой - можно, со стороны сплошной - только при завершении обгона
- Линию 1.2 можно пересекать только для остановки на обочине

Всегда указывай, откуда информация: "Согласно Приложению 2 ПДД..." или "Согласно пункту X ПДД..."
"""

    user_prompt = f"""Контекст для ответа (информация из ПДД):
{context}

Вопрос пользователя: {query}

Ответ:"""

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]

    response = llm.create_chat_completion(
        messages=messages,
        max_tokens=1024,
        temperature=0.3,
        stop=["<|im_end|>"]
    )

    return response["choices"][0]["message"]["content"]
```

```python
def rag_pipeline(query, index, chunks, embedding_model, llm, top_k=5):
    """Полный пайплайн RAG: поиск + генерация"""
    retrieved = retrieve(query, index, chunks, embedding_model, top_k)
    answer = generate_answer(query, retrieved, llm)
    return answer, retrieved
```

# Основная программа

```python
print("=== RAG-бот по ПДД (Qwen2.5-7B) ===")
print("Введите 'exit' или 'quit' для выхода.\n")
print("Примеры вопросов:")
print("  - Какие документы должен иметь водитель?")
print("  - Можно ли обгонять на пешеходном переходе?")
print("  - Что означает знак 3.20?")
print("  - При какой глубине протектора запрещена эксплуатация?")
print("=" * 50 + "\n")

while True:
    user_query = input("Ваш вопрос: ")
    if user_query.lower() in ['exit', 'quit', 'выход']:
        print("До свидания!")
        break

    if not user_query.strip():
        continue

    print("\n🔍 Ищу в ПДД...")
    answer, sources = rag_pipeline(user_query, index, final_chunks, embedding_model, llm, top_k=5)

    print(f"\n📝 Ответ:\n{answer}\n")

    print("📚 Источники:")
    for i, src in enumerate(sources):
        chunk = src['chunk']

        if chunk.get('term'):
            source_info = f"Термин: {chunk['term']}"
        elif chunk.get('chapter'):
            source_info = f"{chunk['chapter']} (пункт {chunk['point']})"
        elif chunk.get('section'):
            source_info = f"{chunk['section']} (пункт {chunk['point']})"
        else:
            source_info = "Информация"

        print(f"  {i+1}. {source_info} (релевантность: {src['score']:.3f})")
    print("\n" + "="*50 + "\n")
```
